#  Data Preparation 
The Adult dataset contains demographic and employment information used to predict whether an individual's annual income exceeds $50,000. During the data preparation phase, missing values represented by the character "?" were identified in the columns workclass, occupation, and native_country. These records were removed using listwise deletion because the proportion of missing observations was relatively small compared with the overall dataset size.

Additional preprocessing steps included removing duplicate records. The cleaned dataset retained more than 30,000 observations, providing a sufficiently large sample for model development and evaluation. These steps improved data quality while maintaining a transparent and reproducible preprocessing workflow.

#  Load the data

```python
from pathlib import Path
import pandas as pd

columns = [
    "age", "workclass", "fnlwgt", "education", "education_num",
    "marital_status", "occupation", "relationship", "race", "sex",
    "capital_gain", "capital_loss", "hours_per_week",
    "native_country", "income"
]

df = pd.read_csv(
    "data/adult.data",
    header=None,
    names=columns,
    skipinitialspace=True,
    na_values="?"
)
df.head()
```

```python
df.info()
```

# Report missing value

```python
missing = df.isnull().sum()

print("missing values by column")
print(missing[missing > 0])

rows_with_missing = df.isna().any(axis=1).sum()

print(f"rows containing at least one missing value: {rows_with_missing:,}")
```

# Remove rows with missing values
# Easy to explain the data
# The data is reproducible 
# No assumptions introduced 
# Still leaves over 30,000 observations

```python
before_rows = len(df)

print(f"# rows before cleaning: {before_rows:,}")

df_clean = df.dropna(ignore_index=True)

print(f"# rows after cleaning: {len(df_clean)}")
```

# Remove duplicates

```python
duplicates = df_clean.duplicated().sum()
print(f"# duplicate rows: {duplicates}")
df_clean = df_clean.drop_duplicates()
print(f"rows after cleaning: {len(df_clean)}")
```

# Final Dataset

```python
print("\nfinal dataset summary")
print(f"original rows: {len(df)}")
print(f"rows removed for missing values: {rows_with_missing:,}")
print(f"duplicate rows removed: {duplicates:,}")
print(f"final rows: {len(df_clean):,}")
```

# Verify target distribution

```python
print(df_clean["income"].value_counts())
```

# Save Sanitized Dataset
Export the cleaned dataset so downstream notebooks can load it directly without re-running the cleaning pipeline.

```python
_out_path = Path("../data/adult_sanitized.csv")

df_clean.to_csv(_out_path, index=False)
print(f"Saved: {_out_path.resolve()}")
print(f"Shape: {df_clean.shape[0]:,} rows × {df_clean.shape[1]} columns")
```
